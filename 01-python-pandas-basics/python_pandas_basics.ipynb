```python
import pandas as pd

url = "https://raw.githubusercontent.com/Dangerbabyxyz/Data-analytics-practice/7fa49ec9da4618aee8ed8a6c12a3462a7cf60252/student_performance.csv"
df = pd.read_csv(url)
df.head()
```

```python
df.shape
df.info()
df.isnull().sum()
df.describe()
```

```python
df["Average"] = df[["Mathematics","English","Chemistry","Physics","Biology"]].mean(axis=1)
df.groupby("Department")["Average"].mean().round(1)
df[["Study_Hours_Weekly","Average"]].corr()
```

```python
print("Rows, columns:", df.shape)
print("\nMissing values:")
print(df.isnull().sum()[df.isnull().sum() > 0])
print("\nAverage score by department:")
print(df.groupby("Department")["Average"].mean().round(1))
print("\nAverage score by gender:")
print(df.groupby("Gender")["Average"].mean().round(1))
print("\nCorrelation, study hours vs average:")
print(round(df["Study_Hours_Weekly"].corr(df["Average"]), 2))
```

```python
for col in ["Attendance_Percent", "English", "Physics"]:
    df[col] = df[col].fillna(df[col].median())

df["Average"] = df[["Mathematics","English","Chemistry","Physics","Biology"]].mean(axis=1)

print("Missing values left:", df.isnull().sum().sum())
print(df[df["Average"] >= 70][["Student", "Average"]].round(1))
```
